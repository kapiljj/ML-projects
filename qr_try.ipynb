```python
import os

mlass_path = '/content/drive/MyDrive/mlass'


first_print_path = os.path.join(mlass_path, 'First Print')
first_print_files = os.listdir(first_print_path)
print("Files in 'First Print':", first_print_files)


second_print_path = os.path.join(mlass_path, 'Second Print')
second_print_files = os.listdir(second_print_path)
print("Files in 'Second Print':", second_print_files)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Function to load images from a folder
def load_images_from_folder(folder):
    images = []
    for filename in os.listdir(folder):
        img = cv2.imread(os.path.join(folder, filename))
        if img is not None:
            images.append(img)
    return images

# Load images from First Print folder
first_print_images = load_images_from_folder(first_print_path)
print(f"Loaded {len(first_print_images)} images from 'First Print'")

# Load images from Second Print folder
second_print_images = load_images_from_folder(second_print_path)
print(f"Loaded {len(second_print_images)} images from 'Second Print'")

# Visualize a sample image from each folder
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(first_print_images[0], cv2.COLOR_BGR2RGB))
plt.title('First Print')
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(second_print_images[0], cv2.COLOR_BGR2RGB))
plt.title('Second Print')
plt.show()
```
